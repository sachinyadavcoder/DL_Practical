```python

# Aim: Implement CBOW Model

import nltk
from gensim.models import Word2Vec

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

# Step 1: Data Preparation
import re
from nltk.corpus import stopwords

text = text.lower()
text = re.sub(r'[^a-z\s]', '', text)

# Step 2: Generate Training Data
sentences = nltk.sent_tokenize(text)
stop_words = set(stopwords.words('english'))

words = []
for sentence in sentences:
    tokens = nltk.word_tokenize(sentence)
    tokens = [w for w in tokens if w not in stop_words]
    if tokens:
        words.append(tokens)

# Step 3: Train CBOW Model
model = Word2Vec(
    words,
    vector_size=50,
    window=2,
    min_count=1,
    sg=0,
    epochs=10
)

# Step 4: Find Similar Words
print("Similar words to alice:")
print(model.wv.most_similar("alice", topn=3))

# Step 5: Predict Output Word
print("\nPredicted output word:")
print(model.predict_output_word(["rabbit", "queen"], topn=3))

# Step 6: Evaluate Model
print("\nWord similarity:")
print(model.wv.similarity("alice", "rabbit"))

print("\nVocabulary size:")
print(len(model.wv))

# Step 7: Save Model
model.save("cbow.model")

print("\nCBOW Model implemented successfully")
```
