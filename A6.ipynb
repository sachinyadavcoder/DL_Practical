```python

# Aim: Object Detection using Transfer Learning

# Step 1: Install and import library
!pip install ultralytics

from ultralytics import YOLO

# Step 2: Load pre-trained YOLOv8 model
model = YOLO("yolov8n.pt")

# Step 3: Freeze lower layers and train detection layers
model.train(
    data="coco8.yaml",
    epochs=2,
    imgsz=320,
    batch=4,
    freeze=10,
    device="cpu"
)

# Step 4: Fine-tune the model by unfreezing layers
model.train(
    data="coco8.yaml",
    epochs=1,
    imgsz=320,
    batch=4,
    freeze=5,
    lr0=0.0001,
    device="cpu"
)

# Step 5: Detect objects in an image
model.predict(
    source="https://ultralytics.com/images/bus.jpg",
    conf=0.25,
    save=True,
    show=True
)

print("Object detection completed successfully!")
```
